```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
os.makedirs("../reports/figures", exist_ok=True)

# 1. Load the processed feature files
tickers = ['GME', 'AMC', 'TSLA', 'FB']
dfs = {}
for t in tickers:
    path = f"../data/processed/features_{t}.parquet"
    if os.path.exists(path):
        dfs[t] = pd.read_parquet(path)
        print(f"Loaded {t}: {len(dfs[t])} posts")

# 2. Plot: Daily Post Volume Over Time per Stock
plt.figure(figsize=(12, 5))
for t, df in dfs.items():
    daily_vol = df.groupby('trade_date').size()
    daily_vol.plot(label=t, alpha=0.8, linewidth=2)

plt.title("Daily WSB Post Volume by Stock (2021)")
plt.xlabel("Trade Date")
plt.ylabel("Number of Posts")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("../reports/figures/eda_daily_volume.png", dpi=300)
plt.show()

# 3. Plot: Sentiment Score Distributions across Stocks
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
metrics = ['sent_pos', 'sent_neu', 'sent_neg']
titles = ['Positive Sentiment', 'Neutral Sentiment', 'Negative Sentiment']

for i, col in enumerate(metrics):
    for t, df in dfs.items():
        sns.kdeplot(df[col], ax=axes[i], label=t, fill=True, alpha=0.2)
    axes[i].set_title(titles[i])
    axes[i].set_xlabel("VADER Polarity Score")
    if i == 0:
        axes[i].set_ylabel("Density")
    axes[i].legend()

plt.tight_layout()
plt.savefig("../reports/figures/eda_sentiment_distribution.png", dpi=300)
plt.show()
```
